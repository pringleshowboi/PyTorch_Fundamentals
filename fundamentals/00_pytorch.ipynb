If we were to add another matrix to our tensor it would have 2 matrices inside it and the shape would be (2, 3, 3)

How would we then reshape and manipulate our data within the Tensor?

```python
TENSOR[0 :] += torch.tensor([1, 1, 1])
TENSOR
```

This only manipulates our 0th dimension of data, how would we append data at the start?

```python
TENSOR = torch.cat((TENSOR, torch.tensor([[[10, 11, 12]]])), dim=1)
TENSOR
```

We appended our data but we want to append it from the start rather than the end we would need follow the rules of torch.cat
When using torch.cat((tensor_A, tensor_B), dim=N), All dimension of tensor_A and tensor_B must match, except for the dimesion specified bt dim=N


Since we want to concatenate along the 0th dimension, the sizes of dim=1 and dim=2 must be identical for both tensors.

To solve this we must change the structure of the data we are passing through to torch.tensor()

```python
new_row_at_start = torch.tensor([[[10, -1, 0]]])
TENSOR = torch.cat((new_row_at_start, TENSOR), dim=1)
TENSOR
```

## Random Tensors
### Why Random Tensors?
- In machine learning, we often want to initialize our model with random weights.
- Random tensors are important because we the way many neural networks learn is that they start with tensors full of random numbers and then adjust those random rumbers to better represent the data

`Start with random numbers -> Look at data -> Update random numbers -> Look at data -> Update random numbers`

```python
# Create random tensor of size/shape (3, 4)
random_tensor = torch.rand(3, 4)
random_tensor
```

```python
random_tensor.shape
```

```python
random_tensor.ndim
```

```python
# Creating a random tensor with similar shape to an image tensor
image_tensor = torch.rand(size=(224, 224, 3)) # height, width, color channels (RGB)
image_tensor.shape, image_tensor.ndim
```

# Creating a range of tensors and tensors-like

```python
#Using torch.arange()
torch.arange(0, 10, step=2)
```

```python
one_to_ten = torch.arange(1, 11)
one_to_ten
```

```python
output_tensor = torch.empty(11)
output_tensor
```

```python
torch.arange(0, 11, out=output_tensor)
output_tensor
```

```python
# using torch.range()
one_to_ten = torch.arange(start=1, end=11, step=1)
one_to_ten
```

```python
# Creating tensors like
ten_zeros = torch.zeros_like(input=one_to_ten)
ten_zeros
```
